## 1. Veille théorique
### 🔹 1. Apprentissage automatique supervisé
L’apprentissage automatique (Machine Learning) consiste à créer des modèles capables d’apprendre à partir des données pour faire des prédictions ou prendre des décisions.
Dans le cas supervisé, les données sont étiquetées c'est à dire qu'on connaît déjà la "bonne réponse".  
**Exemple** : Si on dispose d’un historique de clients ayant ou non souscrit à un dépôt à terme, on peut entraîner un modèle à reconnaître les profils qui ont tendance à dire “oui” à l’offre.
Le but est que le modèle généralise : qu’il fasse de bonnes prédictions sur de nouveaux cas jamais vus.  

### 🔹 2. Données étiquetées (ou labellisées)
Une donnée étiquetée comprend :

des features (colonnes descriptives : âge, emploi, nombre de contacts…)

une target (étiquette) : ici, y = yes ou no.

Ces étiquettes servent à entraîner le modèle : on lui dit ce qu’il doit prédire.  

### 🔹 3. Classification supervisée
La classification est un sous-type d’apprentissage supervisé où la variable cible est catégorielle (classe/famille).

Classe binaire : Oui / Non

Multi-classes : Rouge / Bleu / Vert

Dans notre cas : la classification est binaire → le client souscrit ou ne souscrit pas.

### 🔹 4. Modèle d’apprentissage automatique
Un modèle d’apprentissage automatique est une fonction intelligente qui apprend une relation entre les variables explicatives (features) et la cible.

Par exemple :

Un arbre de décision va poser une série de questions logiques.

Une régression logistique va estimer une probabilité que le client dise oui.

Une fois entraîné, le modèle peut prédire la réponse pour de nouveaux clients.  

### 🔹 5. Donnée cible (target)
La donnée cible (ou label, ou target) est la variable que le modèle cherche à prédire.

### 🔹 6. Données d'entraînement
Pour faire de l'apprentissage automatique on divise notre ensemble de données en deux partie:
  Données d'entrainement   
  Données test  
C’est donc l’ensemble de données qu’on utilise pour entraîner le modèle.
Il contient à la fois les caractéristiques des exemples et les réponses attendues (labels).

le but est de détecter les règles cachées dans les données pour qu’elles soient utilisables ensuite sur de nouveaux cas.  



### 🔹 7. But de la phase d'entraînement
La phase d’entraînement sert à :

Repérer les schémas dans les données d’entrée

dans notre etude il va apprendre la relation existant entre les caractéristiques des clients et leur comportement (souscription ou non)

Le modèle est ajusté automatiquement pour minimiser l’erreur sur les données d'entraînement.

### 🔹 8. But de la phase de prédiction
Une fois entraîné, le modèle est utilisé pour :

Prédire la variable cible (target) sur de nouvelles données sans étiquettes.    

Dans notre cas il va prédire si un nouveau client va souscrire (yes) ou non (no)

Aider les marketers à cibler les bons profils, sans avoir à refaire toute l’analyse manuellement


### 🔹 9. Prétraitement des données (et rôle de PyCaret)
Avant d’entraîner un modèle, il est crucial de :

Supprimer les doublons

Gérer les valeurs manquantes

Encoder les variables catégorielles

Standardiser les valeurs numériques

Pour que les modèles puissent correctement utiliser toutes les données, sans biais ou erreur technique.

PyCaret prend en charge tout cela automatiquement via la fonction setup(), ce qui :

Fait gagner du temps

Réduit les erreurs humaines

Simplifie le pipeline de modélisation  

### 🔹 10. Évaluer un modèle – la métrique accuracy
L’accuracy (ou précision) est une mesure simple mais puissante :

C’est le pourcentage de prédictions correctes sur l’ensemble testé.

Formule : accuracy = (prédictions correctes) / (total des prédictions)
Exemple :
Sur 100 prédictions, si 89 sont correctes → accuracy = 89%.

Attention, l’accuracy peut être trompeuse si les classes sont déséquilibrées (ex : 90% de "non" et 10% de "oui"), d'où l'importance de bien contituer  l'ensemble des données test.

### 🔹 11. Qu’est-ce que l’AutoML ?
L’AutoML (Automated Machine Learning) est une méthode qui :

Gère le prétraitement des données

Automatise le choix des modèles

Compare leurs performances

Il permet de tester des dizaines de modèles très rapidement

C’est ce que propose PyCaret, en une ligne de code (compare_models()).

### 🔹 12. Pourquoi l’AutoML est-il pertinent pour les débutants ?
Parce qu’il :

Masque la complexité du machine learning

Permet d’obtenir des résultats concrets rapidement

Donne une expérience pédagogique progressive

Facilite l’exploration de différents modèles sans expertise avancée

En résumé : un outil parfait pour apprendre sans se noyer dans la technique.

### 🔹 13. Arbre de décision (modèle de classification)
C’est un modèle basé sur une structure en branches et nœuds :

Chaque nœud pose une question (ex : le client est-il marié ?)

Selon la réponse (oui/non), on suit une branche jusqu’à une feuille

La feuille donne la prédiction finale

 Avantages :

Très visuel

Facile à interpréter

Rapide à entraîner
